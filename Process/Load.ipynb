### LOAD TABLAS PRODUCTS, PARTNERS, Y SALES

```python
dbutils.widgets.removeAll()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import functions as F
```

```python
dbutils.widgets.text("catalogo", "catalog_smartdata")
dbutils.widgets.text("esquema_source", "silver")
dbutils.widgets.text("esquema_sink", "golden")
```

```python
catalogo = dbutils.widgets.get("catalogo")
esquema_source = dbutils.widgets.get("esquema_source")
esquema_sink = dbutils.widgets.get("esquema_sink")
```

```python
df_partner_transformed = spark.table(f"{catalogo}.{esquema_source}.partners_transformed")
df_products_transformed = spark.table(f"{catalogo}.{esquema_source}.products_transformed")
df_sales_transformed = spark.table(f"{catalogo}.{esquema_source}.sales_transformed")
```

```python
df_partner_transformed.display()
```

```python
# save the sales table
df_partner_transformed.write.mode("overwrite").saveAsTable(f"{catalogo}.{esquema_sink}.dim_partners")
df_products_transformed.write.mode("overwrite").saveAsTable(f"{catalogo}.{esquema_sink}.dim_products")
df_sales_transformed.write.mode("overwrite").saveAsTable(f"{catalogo}.{esquema_sink}.kpi_sales")
# finished..
```
